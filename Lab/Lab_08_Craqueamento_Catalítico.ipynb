## Contexto

O reator de **FCC (Fluid Catalytic Cracking)** é uma das principais unidades de uma refinaria de petróleo, responsável por converter frações pesadas do petróleo, como gasóleos, em produtos mais leves e de maior valor comercial, como gasolina, GLP e olefinas. O processo ocorre em presença de um catalisador sólido finamente dividido, que se comporta como um fluido devido à alta velocidade do vapor de hidrocarbonetos. No interior do reator, as moléculas pesadas sofrem craqueamento térmico-catalítico em temperaturas elevadas (tipicamente entre 500 e 550 °C), formando compostos menores. Após a reação, o catalisador segue para um regenerador, onde o coque depositado é queimado para restaurar sua atividade, permitindo a operação contínua da unidade.
