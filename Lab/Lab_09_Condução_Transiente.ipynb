## **Modelo Matemático**

O modelo é uma equação diferencial parcial obtida a partir do balanço de energia aplicado a uma barra metálica unidimensional:

$$
\frac{\partial T}{\partial t}=\frac{\partial^2T}{\partial x^2}
\tag{1}
$$

### Condições de Contorno
- Esquerda
$$
T\left(0,t\right)=1
$$

- Direita
$$
\left.\frac{\partial T}{\partial x}\right|_{x=1}=0
$$

### Condição Inicial

$$
T\left(x,0\right)=0
$$

##**🔹 Passo 4 – Definir Método de PVI**

```python
def metodo_Runge_Kutta(t0, y0, n, h):
    # t0 = valor inicial para t
    # y0 = valor inicial para y
    # n = número de passos
    # h = tamanho do passo
    # Requer a função getf(y)

    # Inicializar os vetores de saída para t e y
    t = np.zeros(n)
    y = np.zeros((n,len(y0)))

    # Inicializar x e y com os valores iniciais
    t[0] = t0
    y[0,:] = y0

    # Loop principal para realizar os cálculos
    for i in range(0, n-1):

        # Avaliar as funções de Runge-Kutta
        k1 = getf(y[i,:])
        k2 = getf(y[i,:] + h*k1/2)
        k3 = getf(y[i,:] + h*k2/2)
        k4 = getf(y[i,:] + h*k3)

        # Atualizar y[i+1] pela média ponderada
        y[i+1,:] = y[i,:] + (h/6) *(k1 + 2*k2 + 2*k3 + k4)

        t[i+1] = t[i] + h   # Incrementar o valor de t

    # Retorna os vetores de saída com os valores de t e y calculados
    return t, y
```

##**🔹 Passo 8 – Gerar Vídeo de Simulação**

```python
from matplotlib.animation import FuncAnimation,FFMpegWriter
from IPython.display import HTML

# Subamostragem para reduzir número de frames
nframes = 200

frames = np.linspace(0,T.shape[0]-1,nframes,dtype=int)

# Figura com dois painéis
fig, (ax_bar, ax_plot) = plt.subplots(2, 1,figsize=(10,6),
    gridspec_kw={'height_ratios':[1,3]}
)

# -------------------------
# Painel superior: barra
# -------------------------

im = ax_bar.imshow(T[0,:].reshape(1,-1),cmap='cool',aspect='auto',
                   vmin=T.min(),vmax=T.max())

ax_bar.set_yticks([])
ax_bar.set_xticks([])
ax_bar.set_title("Distribuição de temperatura na barra")

cbar = plt.colorbar(im,ax=ax_bar,orientation='horizontal',pad=0.25)

cbar.set_label('Temperatura')

# -------------------------
# Painel inferior: perfil
# -------------------------

line, = ax_plot.plot(x,T[0,:],lw=3)

ax_plot.set_xlim(x.min(), x.max())
ax_plot.set_ylim(T.min(), T.max())

ax_plot.set_xlabel("Posição")
ax_plot.set_ylabel("Temperatura")

time_text = ax_plot.text(0.02,0.95,'',transform=ax_plot.transAxes)

# -------------------------
# Atualização
# -------------------------

def update(k):

    frame = frames[k]

    # Atualiza barra colorida
    im.set_data(T[frame,:].reshape(1,-1))

    # Atualiza perfil
    line.set_ydata(T[frame,:])

    # Atualiza tempo
    time_text.set_text(f"t = {t[frame]:.2f} s")

    return im, line, time_text

# -------------------------
# Animação
# -------------------------

ani = FuncAnimation(fig,update,frames=len(frames),interval=50,blit=False)

writer = FFMpegWriter(fps=20,bitrate=2000)
ani.save('conducao_calor.mp4',writer=writer)

plt.close()

HTML(ani.to_jshtml())
```
