## Contexto

O escoamento laminar de um fluido newtoniano entre placas paralelas — conhecido como escoamento de Couette–Poiseuille no caso mais geral — é um dos problemas clássicos da mecânica dos fluidos e serve como modelo fundamental para compreender o balanço entre forças viscosas e gradientes de pressão. Esse tipo de configuração aparece em diversas aplicações de engenharia, como no escoamento em canais estreitos, microfluídica, lubrificação hidrodinâmica (filmes de óleo entre superfícies móveis) e em dispositivos de transporte de fluidos em escala microscópica.
